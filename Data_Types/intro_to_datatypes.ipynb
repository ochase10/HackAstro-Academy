## 1.3 Strings

Another important type of data we encounter a lot is text. Python stores these under the type `string`. Note that unlike some other languages, there is no separate data type for characters.

```python
# Strings can be defined with either single or double quotes. They are functionally identical.

hello = 'Hello' 
world = "World"

# As we've seen, some math operations work for strings
print(hello + world)
print(hello * 3)

# Notice that Python doesn't add spaces for you
# Python only adds padding in a couple specific places

# 1) if you pass multiple arguments to print() (using commas to separate them)
print(hello, world)

# 2) it automatically adds a new line character at the end of every print()
print(hello)
print(world)

# You can avoid this with the end option
print(hello, end='') # In this case the character Python adds is the empty string
print(world)
```

Python has numerous built-in methods for manipulating strings. We'll learn many of these over the course of our work, but for now let's see how a couple simple ones work.

```python
s1 = 'Hello, World'

#The .upper() function makes every letter uppercased
print(s1.upper())

#The .lower() function does the opposite
print(s1.lower())
```

Another interesting feature of strings are so called `f-strings` (formatted strings). These allow us to build strings with in-line evaluation, making code cleaner.

```python
# Imagine we wanted to print someone's birth year using their age and the current year

age = 22
current_year = 2026

# This is how we would do that with what we know so far

birth_year = current_year - age
print("You were born in", birth_year)

# With f-strings, we can combine that into 1 line
print(f'You were born in {current_year - age}')


# Even more functionality is found by formatting numbers (i.e. rounding)

pi = 3.14159265358979323846264338

# This is unwieldy to print
print('pi =', pi)

# We can print is as a float with a certain number of decimal places using :#f
print(f'The value of pi is {pi:.2f}')


# Swapping :#f for :#e swaps to scientific notation
big_number = 2876354638272950834092834
print(f"Here's a big number: {big_number:.2e}")

# Notice above that I used "" to enclose the string since I used ' as part of the string itself
```

## Tuples

A tuple is an ordered, immutable container. They are Python's default container for loose collections of items.

More or less, they are un-changeable lists. They are defined using `()` or just commas

```python
tup1 = (1, 2, 3, 4, 5, [3, 2, 4, 5], True, False, 3.454, (1, 4, True, 'String'))

tup2 = 4, [1,2,3], 'hi'

print(type(tup1), type(tup2))
```

```python
# Remember, we can access but not modify elements of a tuple

print(tup2[0])

tup2[0] = 5
```

```python
# Tuples allow us to do funny things like this
a, b = 1, 2

print(a, b, type(a), type(b))
```

# Dictionary

A dictionary is an unordered, mutable collection of key–value pairs. These are basically lists that we access with keys instead of position. Think about a typical dictionary for language. You look up a word and get a longer piece of information associated with it, in that case the definition. In Python, the key and values aren't limited to just text, though. In fact, they can be (almost) anything.

We define them using `{}` and separate entries with `,` and keys/values with `:`.

```python
# There is one item associated with every key; however, that item could be an iterable
# Keys can be (almost) anything but are generally strings
# No particular order
# Values can change and keys can be deleted or added in-place

prices = {'Eggs': 3, 'Milk': 2.50, 'Cereal': 5.99, 'Apples': 4, 'Ice-Cream': 8.97}

# Now we can get the price by just giving the item
print(prices['Cereal'])
```

```python
# Adding a new item to the prices dictionary is easy
# Just assign that element

prices['Bread'] = 

print(prices)
```

```python
# Deleting keys is also quite easy

# del is a keyword used to remove objects from memory, widely applicable in Python
del prices['Eggs']

# We can also remove a key (in place) while returning its associated value
value = prices.pop('Milk')
print(value)

print(prices)
```

## Reference vs Value

```python
# What will happen here?

print(prices)

new = prices

print(new)

new['Juice'] = 12.74

print(prices)
```

### Open after you've run the above cell

Were you surprised by that result? It happened because dictionaries are pass-by-reference (as opposed to pass-by-value). That means that when we do a basic variable assignment, unlike for basic data like ints, we don't make a copy in memory, we just make a pointer to the memory location and assign it to our new variable. So when we modify the "new" dictionary, we modify the old one! This happens with other datatypes, too! Try it out with lists, for example.
